```python
query = """
SELECT TOP 10000
    source_id,
    ra,
    dec,
    phot_g_mean_mag,
    phot_bp_mean_mag,
    phot_rp_mean_mag,
    parallax,
    pmra,
    pmdec
FROM gaiadr3.gaia_source
"""

job = Gaia.launch_job(query)
results = job.get_results()

print("Query successful!")
print("Rows:", len(results))
print("Columns:", results.colnames)
```

```python
df = results.to_pandas()

# Keep a copy of the original data
raw_df = df.copy()

print("Dataset shape:", df.shape)

display(df.head(10))
```

```python
df.info()

print("\nData types:")
print(df.dtypes)
```

```python
missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

display(missing_report)
```

```python
duplicate_count = df["source_id"].duplicated().sum()

print("Duplicate source IDs:", duplicate_count)
print("Total rows:", len(df))
print("Unique source IDs:", df["source_id"].nunique())
```

```python
# Check for infinite numerical values
numeric_df = df.select_dtypes(include=np.number)

print("Infinite values in each column:")
print(np.isinf(numeric_df).sum())

# Check coordinate ranges
invalid_ra = df["ra"].notna() & ~df["ra"].between(0, 360)
invalid_dec = df["dec"].notna() & ~df["dec"].between(-90, 90)

print("\nInvalid right ascension values:", invalid_ra.sum())
print("Invalid declination values:", invalid_dec.sum())
```

```python
df_clean = df.copy()

# Replace infinite values with missing values
df_clean.replace([np.inf, -np.inf], np.nan, inplace=True)

# Remove rows missing essential information
df_clean.dropna(
    subset=["source_id", "ra", "dec"],
    inplace=True
)

# Remove invalid coordinates
valid_ra = df_clean["ra"].between(0, 360)
valid_dec = df_clean["dec"].between(-90, 90)

df_clean = df_clean[valid_ra & valid_dec].copy()

# Remove duplicate source IDs
df_clean.drop_duplicates(
    subset=["source_id"],
    keep="first",
    inplace=True
)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("Remaining duplicate IDs:",
      df_clean["source_id"].duplicated().sum())
```

```python
print("Missing values after cleaning:")
display(df_clean.isnull().sum().to_frame("Missing Count"))

print("\nFinal data types:")
print(df_clean.dtypes)

print("\nFirst 10 cleaned records:")
display(df_clean.head(10))
```

# Task 3: Gaia Dataset — Data Quality Issues and Cleaning Approaches

**Approach:** I used `drop_duplicates(subset=["source_id"], keep="first")` to ensure that each source ID appeared only once. Since no duplicates were present, no rows were removed by this operation.

## 4. Invalid and Infinite Values

I checked the numerical columns for positive and negative infinity and validated the celestial coordinate ranges.

**Approach:** Infinite values were replaced with NaN, and records with invalid coordinates would be removed. However, no infinite values or invalid coordinates were found in this dataset.

## 5. Data Types

## 6. Final Cleaning Results

The original dataset contained 10,000 rows and 9 columns. After cleaning, the dataset still contained 10,000 rows and 9 columns.

The Gaia DR3 dataset was queried, inspected, and cleaned using Python, Astroquery, Astropy, Pandas, and NumPy.
